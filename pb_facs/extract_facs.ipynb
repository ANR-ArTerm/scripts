```python
import xml.etree.ElementTree as ET
from pathlib import Path

CORPUS = Path("../../corpus")
DOSSIERS = ["Architecture", "Peinture", "Perspective"]
OUTPUT_DIR = Path("facs_output")


def extract_facs(xml_path, output_path):
    tree = ET.parse(xml_path)
    root = tree.getroot()

    facs_list = []

    # Avec namespace TEI
    for pb in root.iter('{http://www.tei-c.org/ns/1.0}pb'):
        facs = pb.get('facs')
        if facs:
            facs_list.append(facs)

    # Sans namespace (fallback)
    if not facs_list:
        for pb in root.iter('pb'):
            facs = pb.get('facs')
            if facs:
                facs_list.append(facs)

    if not facs_list:
        print(f"  [!] Aucun facs trouvé dans {xml_path.name}")
        return

    with open(output_path, 'w', encoding='utf-8') as f:
        for facs in facs_list:
            f.write(facs + '\n')

    print(f"  [ok] {xml_path.name} → {output_path} ({len(facs_list)} URL(s))")


def main():
    OUTPUT_DIR.mkdir(exist_ok=True)

    for dossier in DOSSIERS:
        folder = CORPUS / dossier
        if not folder.exists():
            print(f"[!] Dossier introuvable : {folder}")
            continue

        xml_files = list(folder.glob("*.xml"))
        if not xml_files:
            print(f"[!] Aucun XML dans : {folder}")
            continue

        print(f"\n── {dossier} ({len(xml_files)} fichier(s)) ──")
        for xml_file in sorted(xml_files):
            output_file = OUTPUT_DIR / f"{xml_file.stem}.txt"
            extract_facs(xml_file, output_file)

    print(f"\nTerminé. Fichiers TXT dans : {OUTPUT_DIR.resolve()}")


if __name__ == "__main__":
    main()
```

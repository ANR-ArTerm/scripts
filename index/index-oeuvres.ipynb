## Récupération des données et enrichissement des fichiers XML et CSV d'index pour les oeuvres

On cherche ici à récupérer les informations relatives à chaque oeuvre mentionnée dans le corpus à partir de l'URL indiquée dans l'attribut @source de la balise <object>.
Pour chaque oeuvre, on part de 
```<object xml:id="TitreOeuvre" source="URL" type="TypeOeuvre"></object>``` pour arriver à 

```
<object xml:id="TitreOeuvre" source="URL" type="TypeOeuvre">
    <objectIdentifier>
        <objectName>Titre de l'oeuvre</objectName>
        <idno>Identifiant dans l'institution de conservation</idno>
        <institution>Institution de conservation</institution>
    </objectIdentifier>
  <p>Description de l'oeuvre</p>
</object>
```

Ensuite, les entrées ajoutées doivent être classées par ordre alphabétique parmi les entrées existantes. 
Enfin, on met à jour le fichier CSV. 

### Si les infos doivent être récupérées automatiquement, le script est à écrire en s'appuyant sur ceux existant pour les personnes et les lieux.
